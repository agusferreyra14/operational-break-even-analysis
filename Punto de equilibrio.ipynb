```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python

ruta = "C:/Users/Usuario/Desktop/Practica_Datos/PuntoEquilibrioSorter.xlsx"
df = pd.read_excel(ruta)
df.head().round()


X = df[["Bultos Esperados "]]
y = df["Relacion Indución/Clasificación "]
modelo = LinearRegression().fit(X, y)

bultos_equilibrio = (5 - modelo.intercept_) / modelo.coef_[0]
print(f"Punto de equilibrio: {int(bultos_equilibrio)} bultos")
```

```python
x_line = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
y_line = modelo.predict(x_line)

plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', label='Datos reales')
plt.plot(x_line, y_line, color='red', label='Regresión lineal')
plt.axhline(y=5, color='green', linestyle='--', label='Nivel objetivo: 5')
plt.axvline(x=bultos_equilibrio, color='purple', linestyle='--', label=f'Punto de equilibrio ≈ {int(bultos_equilibrio)} bultos')

plt.title("Relación Inducción / Clasificación vs. Bultos Esperados")
plt.xlabel("Bultos Esperados")
plt.ylabel("Relación Inducción / Clasificación")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
